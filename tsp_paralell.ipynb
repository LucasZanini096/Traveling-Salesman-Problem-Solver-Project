```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv('/content/benchmark_results_1748743424.csv')
```

```python
df
```

```python
plt.plot(df['Threads'],df['Execution_Time_s'],color='green')
plt.xticks(df['Threads'])
plt.title("Gráfico de Linha - Threads X Tempo de Execução")
plt.xlabel("Threads")
plt.ylabel("Tempo de Execução (s)")
plt.grid()
plt.show()
```

```python
plt.plot(df['Threads'],df['Speedup'],color='blue')
plt.xticks(df['Threads'])
plt.title("Gráfico de Linha - Threads X Speedup")
plt.xlabel("Threads")
plt.ylabel("Speedup")
plt.grid()
plt.show()
```

```python
plt.plot(df['Threads'],df['Efficiency'],color='orange')
plt.xticks(df['Threads'])
plt.title("Gráfico de Linha - Threads X Eficiência")
plt.xlabel("Threads")
plt.ylabel("Eficiência")
plt.grid()
plt.show()
```

```python
threads = np.arange(len(df['Threads']))
plt.bar(threads - 0.3/2,df['Speedup'],width=0.3,color='blue',label="Speedup")
plt.bar(threads + 0.3/2,df['Efficiency'],width=0.3,color='orange',label="Eficiência")
plt.xticks(threads,df['Threads'])
plt.title("Gráfico de Barra - Threads X Speedup e Eficiência")
plt.xlabel("Threads")
plt.ylabel("Valor")
plt.legend()
plt.grid()
plt.show()
```

```python
plt.plot(df['Threads'],df['Speedup'],color='blue')
plt.plot(df['Threads'],df['Efficiency'],color='orange')
plt.xticks(df['Threads'])
plt.axhspan(1.0, 0, color='green', alpha=0.1)
plt.title("Gráfico de Linha - Threads X Speedup e Eficiência")
plt.xlabel("Threads")
plt.ylabel("Valor")
plt.grid()
plt.show()
```

### Conclusão dos Gráficos/Tabelas

##### Ao estudar o problema e realizar os gráficos acima, podemos perceber que houve um aumento no speedup ao mesmo tempo que houve uma queda na eficiência à medida que o nº de threads aumenta. Esse comportamento, presente nos gráficos, está alinhado com os princípios da Lei de Amdahl (Escalabilidade Forte), a qual trata que a parte sequencial do programa é mantida constante em termos proporcionais ao tamanho do problema, e os gráficos mostram que nem toda adição de threads resulta em uma aceleração proporcional, deixando claro limites do paralelismo imposto por partes não paralelizáveis do programa.
